```python
import pandas as pd

def sql(query):
    return pd.read_sql_query(query, conn)
```

```python
#Q1. Afficher toutes les observations
import sqlite3
import pandas as pd

path = '/content/drive/MyDrive/FMI.xls'

# Tenter de lire en tant que fichier Excel, HTML ou CSV robuste
try:
    # 1. Essai de lecture Excel standard
    df_excel = pd.read_excel(path, engine='xlrd')
except Exception:
    try:
        # 2. Essai de lecture HTML (souvent le cas pour les exports .xls générés par des plateformes)
        dfs = pd.read_html(path)
        df_excel = dfs[0]
    except Exception:
        # 3. Essai de lecture CSV robuste en ignorant les lignes mal formées
        df_excel = pd.read_csv(path, sep=None, engine='python', on_bad_lines='skip')

# Créer la connexion SQLite
conn = sqlite3.connect(':memory:')

# Charger les données dans la table 'fmi_maroc'
df_excel.to_sql('fmi_maroc', conn, index=False, if_exists='replace')

# Exécuter la fonction SQL
display(sql("SELECT * FROM fmi_maroc"))
```

```python
#Q2. Afficher certaines colonnes avec une lecture robuste ligne par ligne
import pandas as pd
import sqlite3

# Lire le fichier brute ligne par ligne pour isoler les données tabulées
rows = []
with open('/content/drive/MyDrive/FMI.xls', 'r', encoding='utf-8', errors='ignore') as f:
    for line in f:
        parts = line.strip().split('\t')
        if len(parts) >= 6 and ('September' in parts[0] or 'December' in parts[0]):
            rows.append(parts)

# Créer le DataFrame nettoyé
df_clean = pd.DataFrame(rows)

# Nommer précisément les colonnes d'après le format standard du FMI pour le Maroc
df_clean.columns = [
    'date', 'quota', 'currency_holdings', 'securities',
    'sdr_holdings', 'outstanding_purchases', 'total_holdings', 'percent_quota'
][:len(df_clean.columns)]

# Recharger proprement dans SQLite
conn = sqlite3.connect(':memory:')
df_clean.to_sql('fmi_maroc', conn, index=False, if_exists='replace')

# Exécuter la requête de vérification pour la Q2
display(sql("""
SELECT date, securities, total_holdings
FROM fmi_maroc
LIMIT 5
"""))
```

```python
#Q6. Compter le nombre total d'observations
display(sql("""
SELECT COUNT(*) AS total_observations
FROM fmi_maroc
"""))
```

```python
#Q7. Calculer la moyenne du total en millions de DTS
display(sql("""
SELECT ROUND(AVG(total_holdings) / 1000000, 1)
AS moyenne_total_mdts
FROM fmi_maroc
"""))
```

```python
#Q8. Trouver le minimum, le maximum et leurs dates
display(sql("""
SELECT
    MIN(total_holdings) AS minimum,
    (SELECT date
     FROM fmi_maroc
     ORDER BY total_holdings ASC
     LIMIT 1) AS date_min,
    MAX(total_holdings) AS maximum,
    (SELECT date
     FROM fmi_maroc
     ORDER BY total_holdings DESC
     LIMIT 1) AS date_max
FROM fmi_maroc
"""))
```

```python
#Q9. Calculer le nombre d'observations et la moyenne par décennie
display(sql("""
SELECT
    (SUBSTR(date, -4, 3) || '0s') AS decennie,
    COUNT(*) AS nb_obs,
    ROUND(AVG(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL)), 1) AS total_moyen_dts
FROM fmi_maroc
WHERE date NOT LIKE '%(in SDRs)%' AND date NOT LIKE '%Date%'
GROUP BY decennie
ORDER BY decennie
"""))
```

```python
#Q10. Afficher les décennies dont la moyenne dépasse 800 millions de DTS
display(sql("""
SELECT
    (SUBSTR(date, -4, 3) || '0s') AS decennie,
    ROUND(AVG(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL)), 1) AS total_moyen_dts
FROM fmi_maroc
WHERE date NOT LIKE '%(in SDRs)%' AND date NOT LIKE '%Date%'
GROUP BY decennie
HAVING AVG(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL)) > 800000000
ORDER BY total_moyen_dts DESC
"""))
```

```python
#Q11. Utiliser les conditions AND et OR avec des colonnes et types corrigés
display(sql("""
SELECT date, securities, sdr_holdings, outstanding_purchases
FROM fmi_maroc
WHERE (
    CAST(REPLACE(REPLACE(securities, ',', ''), ' ', '') AS REAL) >= 100000
    AND CAST(REPLACE(REPLACE(sdr_holdings, ',', ''), ' ', '') AS REAL) < 0
) OR (
    outstanding_purchases IS NOT NULL
    AND CAST(REPLACE(REPLACE(outstanding_purchases, ',', ''), ' ', '') AS REAL) <> 0
)
ORDER BY date
"""))
```

```python
#Q12. Utiliser BETWEEN et IN
display(sql("""
SELECT date, securities, total_holdings
FROM fmi_maroc
WHERE CAST(SUBSTR(date, -4) AS INTEGER) BETWEEN 2000 AND 2010
  AND (SUBSTR(date, -4, 3) || '0s') IN ('2000s', '2010s')
  AND date NOT LIKE '%(in SDRs)%'
  AND date NOT LIKE '%Date%'
ORDER BY CAST(SUBSTR(date, -4) AS INTEGER) ASC, date
"""))
```

```python
#Q13. Classer les 8 plus grandes valeurs du compte n° 1
display(sql("""
SELECT SUBSTR(date, -4) AS annee,
       securities AS nbr1_account,
       RANK() OVER (
           ORDER BY CAST(REPLACE(REPLACE(securities, ',', ''), ' ', '') AS REAL) DESC
       ) AS rang
FROM fmi_maroc
WHERE date NOT LIKE '%(in SDRs)%'
  AND date NOT LIKE '%Date%'
ORDER BY rang
LIMIT 8
"""))
```

```python
#Q14. Afficher les observations supérieures à la moyenne générale
display(sql("""
SELECT date, total_holdings
FROM fmi_maroc
WHERE total_holdings > (
    SELECT AVG(total_holdings)
    FROM fmi_maroc
)
ORDER BY total_holdings DESC
"""))
```

```python
#Q15. Analyser l'évolution des avoirs par décennie après correction du schéma
display(sql("""
SELECT
    (SUBSTR(date, -4, 3) || '0s') AS decennie,
    COUNT(*) AS nb_obs,
    ROUND(AVG(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL)) / 1000000, 1)
    AS total_moyen_mdts,
    ROUND(AVG(100.0 * CAST(REPLACE(REPLACE(securities, ',', ''), ' ', '') AS REAL) /
              NULLIF(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL), 0)), 1)
    AS part_titres_pct,
    ROUND(AVG(100.0 * CAST(REPLACE(REPLACE(sdr_holdings, ',', ''), ' ', '') AS REAL) /
              NULLIF(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL), 0)), 1)
    AS part_sdr_pct,
    ROUND(MAX(CAST(REPLACE(REPLACE(total_holdings, ',', ''), ' ', '') AS REAL)) / 1000000, 1)
    AS max_mdts
FROM fmi_maroc
GROUP BY decennie
ORDER BY decennie
"""))
```
